```python
# ============================================================
# IMPORTAÇÃO DAS BIBLIOTECAS
# ============================================================

import pandas as pd

from pysentimiento import create_analyzer

from google.colab import files
```

```python
# ============================================================
# POLARIZE AI
# Análise de Sentimentos em Comentários do YouTube
# ============================================================

# ---------------- CONFIGURAÇÕES ----------------

# ============================================================
# CARREGAMENTO DA BASE
# ============================================================

from google.colab import files

uploaded = files.upload()

# Obtém automaticamente o nome do arquivo enviado
ARQUIVO_ENTRADA = next(iter(uploaded))

print(f"Arquivo selecionado: {ARQUIVO_ENTRADA}")

# Carrega o arquivo Excel
df = pd.read_excel(ARQUIVO_ENTRADA)

print(f"Total de registros: {len(df)}")

df.head()

COLUNA_TEXTO = "comentario"

ARQUIVO_SAIDA = "AugustoCury02_sentimentos.csv"

# Classes utilizadas pelo modelo
MAPA_SENTIMENTOS = {
    "POS": "Positivo",
    "NEG": "Negativo",
    "NEU": "Neutro"
}
```

```python
# ============================================================
# CARREGAMENTO DA BASE
# ============================================================

df = pd.read_excel(ARQUIVO_ENTRADA)

print(f"Total de registros: {len(df)}")

df.head()
```

```python
# ============================================================
# VERIFICAÇÃO DA BASE
# ============================================================

print("Colunas disponíveis:")
print(df.columns.tolist())
```

```python
# ============================================================
# PREPARAÇÃO DOS TEXTOS
# ============================================================

df[COLUNA_TEXTO] = (
    df[COLUNA_TEXTO]
    .fillna("")
    .astype(str)
)

# Remove comentários vazios
df = df[
    df[COLUNA_TEXTO].str.strip() != ""
].copy()

# Reinicia o índice
df.reset_index(drop=True, inplace=True)

print(f"Registros para análise: {len(df)}")
```

```python
# ============================================================
# CARREGAMENTO DO MODELO
# ============================================================

analyzer = create_analyzer(
    task="sentiment",
    lang="pt"
)

print("Modelo de análise de sentimentos carregado.")
```

```python
# ============================================================
# TESTE DO MODELO
# ============================================================

comentarios_teste = [
    "Esse candidato é muito bom",
    "Não gosto desse político",
    "A entrevista será realizada hoje",
    "Excelente proposta para o Brasil",
    "Esse governo foi péssimo"
]

for texto in comentarios_teste:

    resultado = analyzer.predict(texto)

    print("Comentário:", texto)
    print("Sentimento:", resultado.output)
    print("Probabilidades:", resultado.probas)
    print("-" * 60)
```

```python
# ============================================================
# FUNÇÃO DE ANÁLISE DE SENTIMENTOS
# ============================================================

def analisar_sentimento(texto):

    resultado = analyzer.predict(texto)

    return resultado.output
```

```python
# ============================================================
# ANÁLISE DOS COMENTÁRIOS
# ============================================================

print("Iniciando análise de sentimentos...")

df["sentimento"] = df[COLUNA_TEXTO].apply(
    analisar_sentimento
)

print("Análise concluída!")
```

```python
# ============================================================
# PADRONIZAÇÃO DOS RESULTADOS
# ============================================================

df["sentimento"] = df["sentimento"].map(
    MAPA_SENTIMENTOS
)
```

```python
# ============================================================
# VISUALIZAÇÃO DOS RESULTADOS
# ============================================================

df[
    [
        COLUNA_TEXTO,
        "sentimento"
    ]
].head(20)
```

```python
# ============================================================
# DISTRIBUIÇÃO DOS SENTIMENTOS
# ============================================================

distribuicao = (
    df["sentimento"]
    .value_counts()
)

print(distribuicao)
```

```python
# ============================================================
# PERCENTUAL DOS SENTIMENTOS
# ============================================================

percentual = (
    df["sentimento"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(percentual)
```

```python
# ============================================================
# SENTIMENTO POR CANDIDATO
# ============================================================

sentimento_candidato = pd.crosstab(
    df["candidato"],
    df["sentimento"]
)

sentimento_candidato
```

```python
# ============================================================
# EXPORTAÇÃO DA BASE FINAL
# ============================================================

df.to_csv(
    ARQUIVO_SAIDA,
    index=False,
    encoding="utf-8-sig"
)

print(f"Arquivo salvo: {ARQUIVO_SAIDA}")
```
